```python
FRANCHISES_FEAT_TABLE = "ctl_training_dev.m3.yo_bakehouse_sales_summary"
FRANCHISES_REVIEW_TABLE = "ctl_training_dev.m3.yo_bakehouse_franchise_rating"
FRANCHISES_TABLE = "m3.yo_bakehouse_franchises_single_view"
```

```python
# Read data from feature and review tables
df_feat = spark.table(FRANCHISES_FEAT_TABLE)
df_review = spark.table(FRANCHISES_REVIEW_TABLE)

# Join the two tables on franchiseID
df_joined = df_feat.join(df_review, on="franchiseID", how="inner")

# Write the joined result to the franchises table
df_joined.write.mode("overwrite").saveAsTable(FRANCHISES_TABLE)
```
